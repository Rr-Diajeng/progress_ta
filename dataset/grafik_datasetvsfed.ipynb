```python
import pandas as pd
import matplotlib.pyplot as plt
```

# GHI 1

```python
pred_df = pd.read_csv('./result_fed/GHI_BLN3_0/prediction_inverse.csv')
true_df = pd.read_csv('./dataset_fed/ghi/ghi_dataset.csv')
```

```python
true_df['date_time'] = pd.to_datetime(true_df['date']  + ' '+ true_df['time'])
```

```python
pred_df['date'] = pd.to_datetime(pred_df['date'])
```

```python
true_df.head()
```

```python
start_date = pd.to_datetime('2019-10-01 00:00:00')
```

```python
true_df_filtered = true_df[true_df['date_time'] >= start_date]
pred_df_filtered = pred_df[pred_df['date'] >= start_date]
```

```python
true_df_filtered.shape
```

```python
true_df_filtered.head()
```

```python
pred_df_filtered.shape
```

```python
pred_df_filtered.head()
```

```python
plt.figure(figsize=(12, 8))
```

```python
plt.subplot(3, 1, 1)
plt.plot(pred_df_filtered['date'], pred_df_filtered['KW'], label='CSV PRED KW', color='blue')
plt.plot(true_df_filtered['date_time'], true_df_filtered['KW'], label='CSV TRUE KW', color='orange')
plt.title('KW Comparison')
plt.legend()
```

```python
plt.subplot(3, 1, 2)
plt.plot(pred_df_filtered['date'], pred_df_filtered['TEMPERATURE'], label='CSV PRED TEMPERATURE', color='blue')
plt.plot(true_df_filtered['date_time'], true_df_filtered['TEMPERATURE'], label='CSV TRUE TEMPERATURE', color='orange')
plt.title('TEMPERATURE Comparison')
plt.legend()
```

```python
plt.subplot(3, 1, 2)
plt.plot(pred_df_filtered['date'], pred_df_filtered['OT'], label='CSV PRED GHI', color='blue')
plt.plot(true_df_filtered['date_time'], true_df_filtered['OT'], label='CSV TRUE GHI', color='orange')
plt.title('GHI Comparison')
plt.legend()
```
