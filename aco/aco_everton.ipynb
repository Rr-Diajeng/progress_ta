```python
import numpy as np
```

```python
# Initialize the problem parameters
num_ants = 10  # Number of ants
num_iterations = 50  # Number of iterations
alpha = 1  # Pheromone factor
beta = 5  # Heuristic factor
rho = 0.1  # Pheromone evaporation rate
```

```python
# Define the distance matrix
dist_matrix = np.array([[0, 2, 9, 10],
                       [1, 0, 6, 4],
                       [15, 7, 0, 8],
                       [6, 3, 12, 0]])
```

```python
# Initialize the pheromone matrix
pheromone_matrix = np.ones_like(dist_matrix) / len(dist_matrix)
```

```python
# Initialize the best solution and its cost
best_solution = None
best_cost = np.inf
```

```python
# Perform ant colony optimization
for iteration in range(num_iterations):
    solutions = []
    costs = []

    # Construct solutions for each ant
    for ant in range(num_ants):
        current_city = np.random.randint(len(dist_matrix))
        visited = [current_city]
        solution = [current_city]
        cost = 0

        # Build the solution by selecting the next city based on pheromone and heuristic information
        for _ in range(len(dist_matrix) - 1):
            pheromone_values = pheromone_matrix[current_city] ** alpha
            heuristic_values = (1.0 / dist_matrix[current_city]) ** beta
            probabilities = pheromone_values * heuristic_values
            probabilities[visited] = 0
            probabilities /= probabilities.sum()
            next_city = np.random.choice(len(dist_matrix), p=probabilities)
            visited.append(next_city)
            solution.append(next_city)
            cost += dist_matrix[current_city, next_city]
            current_city = next_city

        # Complete the tour by returning to the starting city
        solution.append(solution[0])
        cost += dist_matrix[current_city, solution[0]]

        # Update the best solution if necessary
        if cost < best_cost:
            best_solution = solution
            best_cost = cost

        solutions.append(solution)
        costs.append(cost)

    # Update the pheromone matrix
    pheromone_matrix *= (1 - rho)
    for solution, cost in zip(solutions, costs):
        for i in range(len(solution) - 1):
            pheromone_matrix[solution[i], solution[i+1]] += 1.0 / cost
```

```python
# Print the best solution and its cost
print("Best solution:", best_solution)
print("Best cost:", best_cost)
```
