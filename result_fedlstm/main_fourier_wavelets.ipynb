# without lstm wavelets

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ghi_df = pd.read_csv('../result_fedlstm/ghi/ghi_withoutlstm_wavelets_percob1.csv', parse_dates=["datetime"])
```

```python
temp_df = pd.read_csv('../result_fedlstm/temperature/temperature_withoutlstm_wavelets_percob3.csv', parse_dates=["datetime"])
```

```python
kw_df = pd.read_csv('../result_fedlstm/kw/kw_withoutlstm_wavelets_percob3.csv', parse_dates=["datetime"])
```

```python
combined_df = pd.DataFrame({
    'datetime': ghi_df['datetime'],
    'ghi': ghi_df['pred_OT'],
    'temperature': temp_df['pred_OT'],
    'kw': kw_df['pred_OT']
})

combined_df.to_csv('1hour_wavelets_withoutlstm_MS.csv', index=False)
```
