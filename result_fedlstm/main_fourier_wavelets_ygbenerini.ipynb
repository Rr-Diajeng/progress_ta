```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ghi_df = pd.read_csv('../result_fedlstm/ghi/corrghi_waveletswithlstm_5.csv', parse_dates=["datetime"])
```

```python
ghi_df['hour'] = ghi_df['datetime'].dt.hour
ghi_df.loc[(ghi_df['hour'] >= 19) | (ghi_df['hour'] <= 5), ['pred_OT', 'true_OT']] = 0
ghi_df.drop(columns=['hour'], inplace=True)
```

```python
temp_df = pd.read_csv('../result_fedlstm/temperature/semuatemp_waveletswithoutlstm_3.csv', parse_dates=["datetime"])
```

```python
kw_df = pd.read_csv('../result_fedlstm/kw/corrkw_waveletswithoutlstm_2.csv', parse_dates=["datetime"])
```

```python
combined_df = pd.DataFrame({
    'datetime': ghi_df['datetime'],
    'ghi': ghi_df['pred_OT'],
    'temperature': temp_df['pred_OT'],
    'kw': kw_df['pred_OT']
})

combined_df.to_csv('dataset_all_fix.csv', index=False)
```
